* Cree una función recursiva que determine si un elemento e está en una lista l

````python
def buscar(lista, elemento, index = 0):
  #caso base
  if(index == len(lista)):
    return False
  #caso recursivo
  if(lista[index] == elemento):
    return True
  buscar(lista, elemento, index+1)

x = [345,345,345,5,5,5,6]
print(buscar(x, 7))
````

* Cree una función recursiva que devuelva la última posición de un número n en una lista l

````python
def ultima_posicion(lista, numero, index=0, posicion = -1):
  if(index == len(lista)):
    return posicion
  if(lista[index] == numero):
    posicion = index
  return ultima_posicion(lista, numero, index + 1, posicion)

x = [4,34,345,5,5]
ultima_posicion(x, 5)
````

````python
def ultima_posicion(lista, numero, index=-1):
  if(index == -len(lista)-1):
    return -1
  if(lista[index] == numero):
    return index + len(lista)
  return ultima_posicion(lista, numero, index - 1)

x = [4,34,345,5,5]
ultima_posicion(x, 4)
````

* Cree una función recursiva que reciba un string s y remueva sus vocales

````python
s = "hola" # --> "hl"
def remover_vocales(string, index = 0, string_sin_vocales = ""):
  vocales = "aeiou"
  if(index == len(string)):
    return string_sin_vocales

  caracter_actual = string[index]
  if(caracter_actual not in vocales):
    string_sin_vocales += caracter_actual
  return remover_vocales(string, index + 1, string_sin_vocales)


remover_vocales(s)
````

````python
s = "hola" # --> "hl"
def remover_vocales(string, index = 0):
  vocales = "aeiou"
  if(index == len(string)):
    return ""

  caracter_actual = string[index]
  if(caracter_actual not in vocales):
    return caracter_actual + remover_vocales(string, index + 1)
  return remover_vocales(string, index+1)

remover_vocales(s)
````

* Cree una función recursiva sin cola que reciba una lista L llena de letras y números y que devuelva una lista con las letras a la izquierda y los números a la derecha

````python
x = [[1,2,3],[5,4,5]]
def recorrer_matriz(matriz, fila=0, columna=0):
  if(fila == len(matriz)):
    return #respuesta(de cola) - parte de la respuesta(sin cola)
  if(columna == len(matriz[fila])):
    return recorrer_matriz(matriz, fila+1, 0)
  print(matriz[fila][columna])
  return recorrer_matriz(matriz, fila, columna+1)

recorrer_matriz(x)
````
